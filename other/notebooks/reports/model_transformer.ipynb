# Load Model

```python
from src.models.sequence_models import TransformerLSTMModel
from src.models.pre_trained_helper import MODEL_NAMES
from src.config import ModelConfig

model_config = ModelConfig()
model_config.model_name = 'Transformer-4-Heads-2-Blocks'

model = TransformerLSTMModel(pre_model_name=MODEL_NAMES[5],
                             device=device,
                             input_dim=1024,
                             linear_hidden_dim=64,
                             num_heads=4,
                             num_blocks=2,
                             dropout=0.6,
                             model_config=model_config,
                             with_lstm=True,
                             lstm_n_layers=2)
```
